```python
import sys
print(sys.executable)

import pandas as pd
print(pd.__version__)
```

```python
from pathlib import Path

DATA = Path("../data")
print(DATA.resolve())
print(sorted(p.name for p in DATA.glob("dvf_2024_*.csv.gz")))
```

```python
# Les codes (département, commune, parcelle…) sont lus comme du texte
# pour ne pas perdre les zéros du début ni casser des valeurs comme "2A".
types_texte = {
    "id_mutation": str,
    "numero_disposition": str,
    "adresse_code_voie": str,
    "code_postal": str,
    "code_commune": str,
    "code_departement": str,
    "ancien_code_commune": str,
    "id_parcelle": str,
    "ancien_id_parcelle": str,
}

morceaux = []
for fichier in sorted(DATA.glob("dvf_2024_*.csv.gz")):
    d = pd.read_csv(fichier, dtype=types_texte)
    print(fichier.name, d.shape)
    morceaux.append(d)

df = pd.concat(morceaux, ignore_index=True)
print("TOTAL :", df.shape)
```

```python
df.head()
```

```python
df.info()
```

- df.isna() marque chaque case vide.
- .mean() donne la part de cases vides par colonne.
- * 100 convertit en pourcentage, et sort_values met les colonnes les plus vides en premier.

```python
vides = (df.isna().mean() * 100).round(1).sort_values(ascending=False)
vides
```

les types de biens vendus:
value_counts compte combien de fois chaque valeur apparaît. dropna=False compte aussi les lignes où le type est vide.

```python
df["type_local"].value_counts(dropna=False)
```

le nombre de ventes différentes:
nunique() compte les valeurs différentes. Si le second chiffre est plus petit que le premier, c’est la preuve que plusieurs lignes appartiennent à la même vente.

```python
print("Lignes :", len(df))
print("Ventes distinctes (id_mutation) :", df["id_mutation"].nunique())
```
